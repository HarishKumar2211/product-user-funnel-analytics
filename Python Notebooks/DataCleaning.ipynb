```python
user_cols = ["visited_website","viewed_product","added_to_cart","checkout_started","purchase_completed","discount_applied"]

for col in user_cols:
    print(col)
    print(df[col].unique())
```

```python
# Fixing Column datatypes

df.dtypes
```

```python
df["date"] = pd.to_datetime(df["date"])
```

```python
# Feature engineering

df[user_cols] = df[user_cols].replace({
    "Yes":1,
     "No":0})

df[user_cols]
```

```python
# Find sessions that added to cart without viewing a product:

df[
    (df["added_to_cart"] == 1) &
    (df["viewed_product"] == 0)
]
```

```python
# Finding sessions with checkout without cart:

df[
    (df["checkout_started"] == 1) &
    (df["added_to_cart"] == 0)
]
```

```python
# Check purchase without checkout:

df[
    (df["purchase_completed"] == 1) &
    (df["checkout_started"] == 0)
]
```

```python
import numpy as np

conditions = [
    df["purchase_completed"] == 1,
    df["checkout_started"] == 1,
    df["added_to_cart"] == 1,
    df["viewed_product"] == 1,
    df["visited_website"] == 1
]

choices = [
    "Purchased",
    "Checkout Started",
    "Added to Cart",
    "Viewed Product",
    "Visited Website"
]

df["funnel_stage"] = np.select(
    conditions,
    choices,
    default="No Activity"
)

df["funnel_stage"].unique()
```

```python
df.head()
```

```python
import numpy as np

conditions = [ 
    df["visited_website"] == 0,
    df["viewed_product"] == 0,
    df["added_to_cart"] == 0,
    df["checkout_started"] == 0,
    df["purchase_completed"] == 0
]

choices = [
    "Website Visit",
    "Product View",
    "Add to Cart",
    "Checkout",
    "Purchase"
]

df["funnel_dropoff_stage"] = np.select(
    conditions,
    choices,
    default="No Drop-off"
)

df.head()
```

```python
df["month_name"] = df["date"].dt.month_name()
```

```python
stage_mapping = {
    "Visited Website": 1,
    "Viewed Product": 2,
    "Added to Cart": 3,
    "Checkout Started": 4,
    "Purchased": 5
}

df["funnel_depth"] = df["funnel_stage"].map(stage_mapping)
```

```python
df.tail()
```

```python
# cart_abandoned

df["cart_abandoned"] = (
    (df["added_to_cart"] == 1) &
    (df["checkout_started"] == 0)
).astype(int)
```

```python
# checkout_abandoned

df["checkout_abandoned"] = (
    (df["checkout_started"] == 1) &
    (df["purchase_completed"] == 0)
).astype(int)
```

```python
# high_intent_session

df["high_intent_session"] = (
    (df["added_to_cart"] == 1) |
    (df["checkout_started"] == 1) |
    (df["purchase_completed"] == 1)
).astype(int)
```

```python
df[["visited_website","viewed_product","added_to_cart","checkout_started","purchase_completed","discount_applied",
    "funnel_stage","funnel_dropoff_stage","funnel_depth","cart_abandoned","checkout_abandoned","high_intent_session"]].tail(10)
```
